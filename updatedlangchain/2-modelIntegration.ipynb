### Streaming

Most models can stream their output content while it is begin generated. By displaying output progressively, streaming singnificantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produces. You can use a loop to process each chunk in real-time.

Streaning means getting the models output token by token (or chunk by chunk) in  real-time, instead of waiting for the complete response. 

Why Steaming Matters:
1. Improves users experience
2. Reduces perceived latency
3. Allows showing intermediate stemps
4. Essential ofr chat applications

```python
model.invoke("Write a 200 words paragraph in Prompt Engineering")
```

```python
for chunk in model.stream("Write a 200 words paragraph in Prompt Engineering"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("why do parrots have colorful feathers?"):
    print(chunk.text, end="|", flush=True)
```

### Batch

Batch allows you to process multiple inputs at once, which is much more efficient than calling .invoke() in a lop

Why use batch?
1. Better performance (Parallelization+fewer API calls)
2. Lower Cost(some providers optimize batch request)
3. Great for processing datasets, multiple questions, etc.

```python
responses = model.batch([
    "why do Narendra Modi not appearing for any press confrence ?",
    "Why people falls in love?",
])
for response in responses:
    print(response)
```

### Tools
Models can request to call tools that perform tasks such as fetchig data from a database, searching the web, or running code. 

#### What is a Tool?
A Tool is basically a funtion with:
1. A clear name
2. A description ( What the LLM reads)
3. Input schema ( what arguments it needs)

Tool Calling Flow

* User asks a question
* LLM decides → "I need to use a tool"
* LLM outputs tool_calls (name + arguments)
* You run the tool → get result
* You pass result back as ToolMessage
* LLM continues reasoning (this loop can repeat)

When we call the model, it can now:
* Decide to call one or more tools
* Return tool calls (Structured JSON) instead of normal text
* You execute the tool and feed the result back


```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response
```

```python
from langchain.tools import  tool

@tool   ## it's an Decorators
def get_weather(location:str)->str:
    """Get the weather at a location"""
    return f"It's Sunny in {location}"

model_with_tools=model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What is the weather like in Boston?")
print(response)

for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```
