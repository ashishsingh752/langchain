### Messages

Messages are the fundamental units of context for model in langchain. They present the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contains:

* Role - Identifies the message type (e.g. system, user)
* content - Represent the actual content of the message (Like text, images, audio, documents, etc.)
* Metadata = Optional fields such as response information, messages IDs, and token usages

Langchain provides a standard messages type that works across all model providers, ensuring considetent behavior regardless of the model begin called. 

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")

```

### Text Prompt

* You don't need conversation history
* You want minimal vode complexity

```python
model.invoke("What is the Aritificial Intelligence?")
```

### Messages Prompt

We can also pass in a list of messages to the model by providing a list of messages objects. 

Message Types
* System Message- Tells the model how to behave and provide context for interactions
* Human Message- Represent user input and interactions with the model
* AI Message- Response generated by the model, including text content, tool calls, and metadata
* Tool message- Represent the output of the tool calls

#### System Message
Role: System \
This is the most important message at the beginning. It sets the behavior, personality, rules, and instruntion for the AI. \
It's like instruntion from the developer to the AI model. \

#### Human Message
Role: Human \
It represents user input and interactions. They can contain text, images, audio, file and any other amount of multimodal content. \ 

#### AI messages
Role: assistant \
It represents the output of a model invocation. They can include multimodal data, tool calls, and provider-specific metadata that you can later access. Or we can say that it represents what the AI(Model) responds with. 

#### Tool message
Role: Tool
It contains the result of a tool that was called by the AI. It is used only when we are working with the tools/Agents.


```python
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage,
    ToolMessage
)


messages = [
    SystemMessage(content="You are a helpful AI tutor for Python."),
    HumanMessage(content="What is a list comprehension?")
]

response = model.invoke(messages)
print(response.content)
print(response)
```

```python
messages = [
    SystemMessage(content="You are a friendly history teacher."),
    HumanMessage(content="Who was the first President of India?"),
    AIMessage(content="Dr. Rajendra Prasad was the first President of India."),
    HumanMessage(content="When did he take office?"),
]

response = model.invoke(messages)
print(response.usage_metadata)
```

```python
from langchain.messages import (
    AIMessage, 
    HumanMessage,
    ToolMessage,
    SystemMessage
)

ai_message=AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"Delhi"},
        "id":"123_call"
    }]
)

weather_result = "Sunny, 40 C"

tool_message=ToolMessage(
    content=weather_result,
    tool_call_id="123_call"
)

messages=[
    HumanMessage("What is the weather in the Lucknow"),
    ai_message,
    tool_message
]

response=model.invoke(messages)

```
